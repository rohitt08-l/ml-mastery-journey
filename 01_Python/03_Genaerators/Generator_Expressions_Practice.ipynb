## Practice 1: Basic Generator Expression

```python
gen = (x*x for x in range(5))
print(gen)
```

## Practice 2: Iterate Through Generator

```python
gen = (x*x for x in range(5))
for value in gen:
    print(value)
```

## Practice 3: Using next()

```python
gen = (x*x for x in range(5))
print(next(gen))
print(next(gen))
print(next(gen))
```

## Practice 4: Conditional Generator

```python
evens = (x for x in range(20) if x % 2 == 0)
print(list(evens))
```

## Practice 5: Word Length Generator

```python
words = ['apple','banana','kiwi']
lengths = (len(word) for word in words)
print(list(lengths))
```

## Practice 6: Memory Comparison

```python
import sys
lst = [x for x in range(1000)]
gen = (x for x in range(1000))
print('List:', sys.getsizeof(lst))
print('Generator:', sys.getsizeof(gen))
```
