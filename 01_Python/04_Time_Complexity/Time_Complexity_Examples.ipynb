# Time Complexity Examples Notebook

Learn each complexity with simple Python examples.

## O(1) - Constant Time

```python
arr=[10,20,30,40,50]
print(arr[2])
```

## O(log n) - Binary Search

```python
arr=list(range(1,17))
target=13
low,high=0,len(arr)-1
while low<=high:
    mid=(low+high)//2
    if arr[mid]==target:
        print('Found',mid)
        break
    elif arr[mid]<target:
        low=mid+1
    else:
        high=mid-1
```

## O(n) - Linear Traversal

```python
nums=[1,2,3,4,5]
total=0
for x in nums:
    total+=x
print(total)
```

## O(n log n) - Sorting

```python
nums=[5,2,8,1,9,3]
nums.sort()
print(nums)
```

## O(n²) - Nested Loops

```python
for i in [1,2,3]:
    for j in [1,2,3]:
        print(i,j)
```

## O(n³) - Triple Nested Loops

```python
count=0
n=3
for i in range(n):
    for j in range(n):
        for k in range(n):
            count+=1
print(count)
```

## O(2^n) - Recursive Fibonacci

```python
def fib(n):
    if n<=1:
        return n
    return fib(n-1)+fib(n-2)
print(fib(6))
```

## O(n!) - Permutations

```python
from itertools import permutations
for p in permutations([1,2,3]):
    print(p)
```
