# Dict Comprehensions

Dictionary comprehension is used to create a dictionary in a short and clear way. It allows keys and values to be generated from a loop in one line. This helps in building dictionaries directly without writing multiple statements.

Baasic Syntax : {key:value for item in iterable}

```python
a={x: x**2 for x in range(3)}
a
```

```python
#creating dictionary from 2 lists
keys = ['a','b','c','d','e']
values = [1, 2, 3, 4, 5]  
dict1={k:v for k,v in zip(keys,values)}
dict1
```

Using fromkeys() Method

The fromkeys() method creates a dictionary by taking a group of keys and assigning the same value to all of them.

```python
d = dict.fromkeys(range(5), True)
print(d)
```

```python
#conditional dict comprehension
evens = {x:x*x for x in range(10) if x%2==0}
print(evens)
```

Questions solving

```python
#Q1. Create a dictionary of numbers and their squares
squares={x:x**2 for x in range(5)}
print(squares)
```

```python
#Q2. Create a dictionary of numbers and their cubes
cubes={x:x**3 for x in range(5)}
print(cubes)
```

```python
#Q3. Create a dictionary where keys are numbers and values are double the number
dobles={x:2*x for x in range(5)}
print(dobles)
```

```python
#Q4. Create a dictionary with words and their lengths
words = ["apple", "banana", "cherry"]
word_lengths ={word:len(word) for word in words}
print(word_lengths)
```

```python
#Q5. Create a dictionary of characters and their ASCII values
chars = ['a', 'b', 'c']
ascii_values = {char:ord(char) for char in chars}
print(ascii_values)
```

```python
#Q6. Create a dictionary of numbers and whether they are even or odd
numbers = [1, 2, 3, 4, 5]
even_odd = {num:'Even' if num%2==0 else 'Odd' for num in numbers}
print(even_odd)
```

```python
#7. Create a dictionary containing only words with length greater than 5
words = ["apple", "banana", "watermelon", "kiwi"]
words_new={word:len(word) for word in words if len(word)>5}
print(words_new)
```

```python
#8 Swap keys and values in a dictionary
data = {'a':1, 'b':2, 'c':3}
swapped={v:k for k,v in data.items()}
print(swapped)
```

```python
#9 Degree to Fahrenheit
temps = [0, 10, 20, 30]
fahrenheit = {c: (9/5)*c + 32 for c in temps}
print(fahrenheit)
```
