```python
word = "banana"
chars = {char for char in word}
print(chars)
```

```python
names = ["rohit", "amit", "rohit"]
upper = {name.upper() for name in names}
print(upper)
```

```python
chars = ['a', 'b', 'c']
ascii_values = {ord(char) for char in chars}
print(ascii_values)
```

```python
words = ["apple", "banana", "watermelon", "kiwi"]
words1={word for word in words if len(word)>5}
print(words1)
```

```python
data = ["apple", "", "banana", "", "kiwi"]
non_empty={word for word in data if len(word)>0}
print(non_empty)
```
