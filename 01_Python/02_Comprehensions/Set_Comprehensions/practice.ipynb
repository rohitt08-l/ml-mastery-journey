```python
squares = {x*x for x in range(1, 6)}
squares
```

```python
cubes = {x**3 for x in range(1, 6)}
print(cubes)
```

```python
word = "banana"
chars = {char for char in word}
print(chars)
```

```python
names = ["rohit", "amit", "rohit"]
upper = {name.upper() for name in names}
print(upper)
```

```python
chars = ['a', 'b', 'c']
ascii_values = {ord(char) for char in chars}
print(ascii_values)
```
