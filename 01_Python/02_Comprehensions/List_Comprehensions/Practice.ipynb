List Comprehensions

###### List comprehension is a compact and elegant way to create lists in Python.

Basic syntax
[expression for item in iterable]

```python
result = [i for i in range(5)]
result
```

```python
#Example 1 — Squares of Numbers
squares =[i**2 for i in range(5)]
squares
```

```python
#Example 2 — Convert Strings to Uppercase
names = ["rohit", "amit", "sneha"]
uppercase_names = [name.upper() for name in names]
uppercase_names
```

```python
# Example 3 — Find Length of Words
words = ["apple", "banana", "cherry"]
word_len = [len(word) for word in words]
word_len
```

```python
#Example 4 — Postive or Negative
numbers = [-2, -1, 0, 1, 2]
pos=["positive" if num >0 else "negative" if num <0 else "zero" for num in numbers]
pos
```

```python
#Example 5 - Find Common Elements between Two Lists
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]
common = [num for num in list1 if num in list2]
common
```

```python
# Example 6 : Extract characters from a String starting with a Vowel
words = ["apple", "banana", "orange", "grapes"]
vowels = 'aeiou'
vowels_words=[word for word in words if word[0].lower() in vowels]
vowels_words
```

```python
#Example 7 - Replace Negative Numbers from a List with  0
nums = [2, -1, 5, -7, 8]
pos=[num if num>0 else 0 for num in nums]
pos
```
